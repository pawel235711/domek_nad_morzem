## 1. PARAMETRY WEJŚCIOWE

### 1.1. Wymiary budynku

```python
# ============================================================
# WYMIARY BUDYNKU - MODYFIKUJ TUTAJ
# ============================================================

L = 8.00        # Długość budynku [m]
W = 3.50        # Szerokość budynku [m]
H_total = 4.65  # Wysokość całkowita budynku [m]
H_parter = 2.30 # Wysokość parteru w świetle [m]
H_antresola = 1.90  # Wysokość antresoli w świetle [m]

# Antresola
L_ant = 3.00    # Długość antresoli [m]
W_ant = 3.00    # Szerokość antresoli [m]

# Łazienka
L_laz = 2.00    # Długość łazienki [m]
W_laz = 1.50    # Szerokość łazienki [m]

print(f"Powierzchnia zabudowy: {L * W:.1f} m²")
print(f"Powierzchnia antresoli: {L_ant * W_ant:.1f} m²")
print(f"Wysokość budynku: {H_total:.2f} m")
```

### 1.2. Parametry konstrukcji stalowej

```python
# ============================================================
# KONSTRUKCJA STALOWA - MODYFIKUJ TUTAJ
# ============================================================

# Profil stalowy zamknięty kwadratowy
profil_wymiar = 100     # Wymiar profilu [mm]
profil_grubosc = 4      # Grubość ścianki [mm]

# Masa liniowa profili [kg/m] - z tablic producenta
masa_profilu_100x100x4 = 11.6  # kg/m
masa_profilu_100x100x3 = 8.96  # kg/m

# Właściwości mechaniczne stali S235JR
fy = 235        # Granica plastyczności [MPa]
gamma_M0 = 1.0  # Współczynnik bezpieczeństwa materiału
gamma_M1 = 1.1  # Współczynnik bezpieczeństwa na wyboczenie

# Właściwości geometryczne profilu 100x100x4
A = 14.8        # Pole przekroju [cm²]
W_pl = 35.33    # Wskaźnik wytrzymałości plastycznej [cm³]
i = 3.72        # Promień bezwładności [cm]

print(f"Profil: {profil_wymiar}x{profil_wymiar}x{profil_grubosc} mm")
print(f"Stal: S235JR (fy = {fy} MPa)")
print(f"Pole przekroju: {A} cm²")
print(f"Wskaźnik wytrzymałości: {W_pl} cm³")
```

### 1.3. Obciążenia

```python
# ============================================================
# OBCIĄŻENIA - MODYFIKUJ TUTAJ
# ============================================================

# Obciążenia stałe [kN/m²]
q_dach_stale = 0.65      # Ciężar własny dachu
q_strop_stale = 0.75     # Ciężar własny stropu antresoli
q_sciany = 0.50          # Ciężar własny ścian

# Obciążenia zmienne użytkowe [kN/m²]
q_strop_uzytkowe = 2.0   # Obciążenie użytkowe stropu (ludzie, meble)
q_dach_uzytkowe = 0.75   # Obciążenie użytkowe dachu (konserwacja)

# Obciążenia klimatyczne [kN/m²]
q_snieg = 0.96           # Obciążenie śniegiem (strefa III)
q_wiatr = 0.65           # Obciążenie wiatrem (strefa I)

# Współczynniki bezpieczeństwa
gamma_G = 1.35           # Dla obciążeń stałych
gamma_Q = 1.50           # Dla obciążeń zmiennych

print("Obciążenia stałe:")
print(f"  Dach: {q_dach_stale} kN/m²")
print(f"  Strop: {q_strop_stale} kN/m²")
print(f"\nObciążenia zmienne:")
print(f"  Użytkowe (strop): {q_strop_uzytkowe} kN/m²")
print(f"  Śnieg: {q_snieg} kN/m²")
```

### 1.4. Dane materiałowe (dla obliczeń masy)

```python
# ============================================================
# DANE MATERIAŁOWE - MODYFIKUJ TUTAJ
# ============================================================

# Masy jednostkowe materiałów [kg/m²]
masa_osb_18 = 10.8      # OSB 18 mm
masa_osb_22 = 13.2      # OSB 22 mm
masa_osb_20 = 12.0      # OSB 20 mm
masa_osb_12 = 7.2       # OSB 12 mm
masa_welny_10cm = 3.5   # Wełna mineralna 10 cm
masa_blachy = 4.0       # Blacha na rąbek 0,5 mm
masa_boazerii = 7.5     # Deski elewacyjne 15 mm
masa_paneli = 7.0       # Panele podłogowe 8 mm
masa_papy = 5.0         # Papa termozgrzewalna
masa_folii = 0.2        # Folia paroizolacyjna
masa_plytek = 20.0      # Płytki ceramiczne z klejem

print("Masy jednostkowe materiałów:")
print(f"  OSB 18 mm: {masa_osb_18} kg/m²")
print(f"  Wełna 10 cm: {masa_welny_10cm} kg/m²")
print(f"  Blacha: {masa_blachy} kg/m²")
```

## 2. OBLICZENIA KONSTRUKCYJNE

### 2.1. Sprawdzenie belki stropu antresoli

```python
# ============================================================
# BELKA STROPU ANTRESOLI
# ============================================================

print("=" * 70)
print("SPRAWDZENIE BELKI STROPU ANTRESOLI")
print("=" * 70)

# Parametry belki
L_belka = L_ant          # Rozpiętość belki [m]
rozstaw_belek = 0.60     # Rozstaw belek [m]

# Obciążenie na belkę
q_stale_belka = q_strop_stale * rozstaw_belek
q_zmienne_belka = q_strop_uzytkowe * rozstaw_belek
q_oblicz = gamma_G * q_stale_belka + gamma_Q * q_zmienne_belka

print(f"\nRozpiętość belki: {L_belka:.2f} m")
print(f"Rozstaw belek: {rozstaw_belek:.2f} m")
print(f"\nObciążenie stałe na belkę: {q_stale_belka:.2f} kN/m")
print(f"Obciążenie zmienne na belkę: {q_zmienne_belka:.2f} kN/m")
print(f"Obciążenie obliczeniowe: {q_oblicz:.2f} kN/m")

# Moment zginający maksymalny (belka swobodnie podparta)
M_Ed = (q_oblicz * L_belka**2) / 8

print(f"\nMaksymalny moment zginający M_Ed = {M_Ed:.2f} kNm")

# Nośność na zginanie
M_c_Rd = (W_pl * fy) / 1000  # Konwersja: cm³ * MPa = Nm / 1000 = kNm

print(f"Nośność na zginanie M_c,Rd = {M_c_Rd:.2f} kNm")

# Sprawdzenie warunku
wykorzystanie = (M_Ed / M_c_Rd) * 100
zapas = M_c_Rd / M_Ed

print(f"\nWykorzystanie nośności: {wykorzystanie:.1f}%")
print(f"Zapas nośności: {zapas:.2f}x")

if M_Ed <= M_c_Rd:
    print("\n✓ WARUNEK SPEŁNIONY - belka jest bezpieczna!")
else:
    print("\n✗ WARUNEK NIE SPEŁNIONY - belka wymaga wzmocnienia!")
```

### 2.2. Sprawdzenie słupa nośnego

```python
# ============================================================
# SŁUP NOŚNY
# ============================================================

print("=" * 70)
print("SPRAWDZENIE SŁUPA NOŚNEGO")
print("=" * 70)

# Powierzchnia zbierania obciążeń przez słup narożny
A_slup = (L / 2) * (W / 2)

print(f"\nPowierzchnia zbierania obciążeń: {A_slup:.2f} m²")

# Siła od dachu
N_dach = (q_dach_stale + q_snieg) * A_slup

# Siła od stropu antresoli (tylko jeśli słup pod antresolą)
A_slup_ant = min(A_slup, L_ant * W_ant / 4)  # Ograniczone do antresoli
N_strop = (q_strop_stale + q_strop_uzytkowe) * A_slup_ant

print(f"\nSiła od dachu (stałe + śnieg): {N_dach:.2f} kN")
print(f"Siła od stropu (stałe + użytkowe): {N_strop:.2f} kN")

# Siła obliczeniowa
N_Ed = gamma_G * N_dach + gamma_Q * N_strop

print(f"\nSiła ściskająca obliczeniowa N_Ed = {N_Ed:.2f} kN")

# Nośność na ściskanie (bez wyboczenia)
N_c_Rd = (A * fy) / gamma_M0 / 10  # Konwersja: cm² * MPa / 10 = kN

print(f"\nNośność na ściskanie N_c,Rd = {N_c_Rd:.0f} kN")

# Sprawdzenie wyboczenia
L_cr = H_total  # Długość wyboczeniowa [m]
lambda_1 = 93.9 * (235 / fy)**0.5  # Smukłość graniczna
lambda_slup = (L_cr * 100) / i  # Smukłość słupa
lambda_wzgl = lambda_slup / lambda_1  # Smukłość względna

print(f"\nDługość wyboczeniowa: {L_cr:.2f} m")
print(f"Smukłość słupa λ = {lambda_slup:.1f}")
print(f"Smukłość względna λ̄ = {lambda_wzgl:.2f}")

# Współczynnik wyboczeniowy (uproszczone dla krzywej c)
alpha = 0.49  # Dla krzywej wyboczeniowej c
phi = 0.5 * (1 + alpha * (lambda_wzgl - 0.2) + lambda_wzgl**2)
chi = min(1.0, 1 / (phi + (phi**2 - lambda_wzgl**2)**0.5))

print(f"Współczynnik wyboczeniowy χ = {chi:.2f}")

# Nośność na wyboczenie
N_b_Rd = (chi * A * fy) / gamma_M1 / 10

print(f"\nNośność na wyboczenie N_b,Rd = {N_b_Rd:.0f} kN")

# Sprawdzenie warunku
wykorzystanie_wyboczenie = (N_Ed / N_b_Rd) * 100
zapas_wyboczenie = N_b_Rd / N_Ed

print(f"\nWykorzystanie nośności: {wykorzystanie_wyboczenie:.1f}%")
print(f"Zapas nośności: {zapas_wyboczenie:.1f}x")

if N_Ed <= N_b_Rd:
    print("\n✓ WARUNEK SPEŁNIONY - słup jest bezpieczny!")
else:
    print("\n✗ WARUNEK NIE SPEŁNIONY - słup wymaga wzmocnienia!")
```

## 3. OBLICZENIA MASY CAŁKOWITEJ

### 3.1. Konstrukcja stalowa

```python
# ============================================================
# MASA KONSTRUKCJI STALOWEJ
# ============================================================

print("=" * 70)
print("MASA KONSTRUKCJI STALOWEJ")
print("=" * 70)

# Rama fundamentowa
rama_fund_obwod = 2 * (L + W)
n_belek_podloga = int(L / 1.0) - 1
dlug_belek_podloga = W * n_belek_podloga
rama_fund_total = rama_fund_obwod + dlug_belek_podloga
masa_rama_fund = rama_fund_total * masa_profilu_100x100x4

print(f"\nRama fundamentowa: {rama_fund_total:.1f} m × {masa_profilu_100x100x4} kg/m = {masa_rama_fund:.1f} kg")

# Słupy
n_slupow = 8
dlug_slupow = n_slupow * H_total
masa_slupow = dlug_slupow * masa_profilu_100x100x3

print(f"Słupy ({n_slupow} szt.): {dlug_slupow:.1f} m × {masa_profilu_100x100x3} kg/m = {masa_slupow:.1f} kg")

# Rama dachowa
rama_dach_total = rama_fund_total  # Taka sama jak fundamentowa
masa_rama_dach = rama_dach_total * masa_profilu_100x100x4

print(f"Rama dachowa: {rama_dach_total:.1f} m × {masa_profilu_100x100x4} kg/m = {masa_rama_dach:.1f} kg")

# Strop antresoli
rama_strop_obwod = 2 * (L_ant + W_ant)
n_belek_strop = int(L_ant / rozstaw_belek) - 1
dlug_belek_strop = W_ant * n_belek_strop
podciag = L_ant
strop_total = rama_strop_obwod + dlug_belek_strop + podciag
masa_strop = strop_total * masa_profilu_100x100x4

print(f"Strop antresoli: {strop_total:.1f} m × {masa_profilu_100x100x4} kg/m = {masa_strop:.1f} kg")

# Stężenia
dlug_stezen = 20.0
masa_stezen = dlug_stezen * 3.0

print(f"Stężenia: {masa_stezen:.1f} kg")

masa_stal_total = masa_rama_fund + masa_slupow + masa_rama_dach + masa_strop + masa_stezen

print(f"\n>>> ŁĄCZNA MASA KONSTRUKCJI STALOWEJ: {masa_stal_total:.1f} kg")
```

### 3.2. Płyty OSB

```python
# ============================================================
# MASA PŁYT OSB
# ============================================================

print("=" * 70)
print("MASA PŁYT OSB")
print("=" * 70)

# Ściany
obwod_scian = 2 * (L + W)
pow_scian = obwod_scian * H_total
pow_okien = (2.0*2.1) + (1.2*0.9) + (0.6*0.9) + (1.5*0.9) + (0.9*2.1)
pow_scian_netto = pow_scian - pow_okien
masa_osb_sciany = pow_scian_netto * masa_osb_18

print(f"\nŚciany (OSB 18 mm): {pow_scian_netto:.1f} m² × {masa_osb_18} kg/m² = {masa_osb_sciany:.1f} kg")

# Dach
pow_dachu = L * W
masa_osb_dach = pow_dachu * masa_osb_20 * 2

print(f"Dach (2× OSB 20 mm): {pow_dachu:.1f} m² × {masa_osb_20*2} kg/m² = {masa_osb_dach:.1f} kg")

# Podłoga
masa_osb_podloga = pow_dachu * masa_osb_22

print(f"Podłoga (OSB 22 mm): {pow_dachu:.1f} m² × {masa_osb_22} kg/m² = {masa_osb_podloga:.1f} kg")

# Strop antresoli
pow_stropu = L_ant * W_ant
masa_osb_strop = pow_stropu * masa_osb_22 * 2

print(f"Strop antresoli (2× OSB 22 mm): {pow_stropu:.1f} m² × {masa_osb_22*2} kg/m² = {masa_osb_strop:.1f} kg")

# Osłona spodnia
masa_osb_spod = pow_dachu * masa_osb_12

print(f"Osłona spodnia (OSB 12 mm): {pow_dachu:.1f} m² × {masa_osb_12} kg/m² = {masa_osb_spod:.1f} kg")

masa_osb_total = masa_osb_sciany + masa_osb_dach + masa_osb_podloga + masa_osb_strop + masa_osb_spod

print(f"\n>>> ŁĄCZNA MASA PŁYT OSB: {masa_osb_total:.1f} kg")
```

### 3.3. Pozostałe materiały

```python
# ============================================================
# MASA POZOSTAŁYCH MATERIAŁÓW
# ============================================================

print("=" * 70)
print("MASA POZOSTAŁYCH MATERIAŁÓW")
print("=" * 70)

# Izolacja
masa_welna_sciany = pow_scian_netto * masa_welny_10cm
masa_welna_dach = pow_dachu * masa_welny_10cm
masa_welna_podloga = pow_dachu * masa_welny_10cm
masa_welna_strop = pow_stropu * (masa_welny_10cm / 2)
masa_welna_total = masa_welna_sciany + masa_welna_dach + masa_welna_podloga + masa_welna_strop

print(f"\nIzolacja (wełna): {masa_welna_total:.1f} kg")

# Elewacja
masa_blacha_total = pow_scian_netto * masa_blachy

print(f"Elewacja (blacha): {masa_blacha_total:.1f} kg")

# Pokrycie dachu
masa_papa_total = pow_dachu * masa_papy

print(f"Pokrycie dachu (papa): {masa_papa_total:.1f} kg")

# Wykończenie wewnętrzne
pow_scian_wewn = obwod_scian * H_parter + 2 * (L_ant + W_ant) * H_antresola - pow_okien
pow_sufit = pow_dachu
pow_boazerii = pow_scian_wewn + pow_sufit
masa_boazeria_total = pow_boazerii * masa_boazerii

pow_paneli = pow_dachu + pow_stropu - (L_laz * W_laz)
masa_panele_total = pow_paneli * masa_paneli

pow_plytek = (L_laz * W_laz) + 2 * (L_laz + W_laz) * 2.0
masa_plytki_total = pow_plytek * masa_plytek

masa_wykonczenie_total = masa_boazeria_total + masa_panele_total + masa_plytki_total

print(f"Wykończenie wewnętrzne: {masa_wykonczenie_total:.1f} kg")

# Folie
pow_folii = pow_scian_netto + pow_dachu + pow_dachu
masa_folie_total = pow_folii * masa_folii * 2

print(f"Folie i membrany: {masa_folie_total:.1f} kg")

# Stolarka
masa_stolarka_total = 35 + 80 + 25 + 12 + 30 + 15

print(f"Stolarka okienna i drzwiowa: {masa_stolarka_total:.0f} kg")

# Schody
masa_schodow = 45

print(f"Schody: {masa_schodow:.0f} kg")

# Balustrada
masa_balustrady = (L_ant + W_ant) * 8

print(f"Balustrada: {masa_balustrady:.1f} kg")

# Obróbki
masa_obrobki = 40

print(f"Obróbki blacharskie: {masa_obrobki:.0f} kg")

# Instalacje
masa_instalacje_total = 30 + 40 + 30 + 80

print(f"Instalacje: {masa_instalacje_total:.0f} kg")
```

### 3.4. Podsumowanie masy całkowitej

```python
# ============================================================
# PODSUMOWANIE MASY CAŁKOWITEJ
# ============================================================

import pandas as pd

print("=" * 70)
print("PODSUMOWANIE MASY CAŁKOWITEJ BUDYNKU")
print("=" * 70)

kategorie = [
    ("Konstrukcja stalowa", masa_stal_total),
    ("Płyty OSB", masa_osb_total),
    ("Izolacja (wełna mineralna)", masa_welna_total),
    ("Elewacja (blacha na rąbek)", masa_blacha_total),
    ("Pokrycie dachu (papa)", masa_papa_total),
    ("Wykończenie wewnętrzne", masa_wykonczenie_total),
    ("Folie i membrany", masa_folie_total),
    ("Stolarka okienna i drzwiowa", masa_stolarka_total),
    ("Schody Dolle Normandie", masa_schodow),
    ("Balustrada antresoli", masa_balustrady),
    ("Obróbki blacharskie, rynny", masa_obrobki),
    ("Instalacje i wyposażenie", masa_instalacje_total),
]

masa_calkowita = sum([m for _, m in kategorie])

# Tworzenie tabeli
df = pd.DataFrame(kategorie, columns=["Kategoria", "Masa [kg]"])
df["Udział [%]"] = (df["Masa [kg]"] / masa_calkowita * 100).round(1)

print("\n" + df.to_string(index=False))
print("\n" + "-" * 70)
print(f"MASA CAŁKOWITA BUDYNKU (bez mebli): {masa_calkowita:.1f} kg")
print(f"Masa całkowita: {masa_calkowita/1000:.2f} tony")
print(f"Masa na m² zabudowy: {masa_calkowita/(L*W):.1f} kg/m²")

# Obciążenie na fundamenty
print(f"\nOBCIĄŻENIE NA FUNDAMENTY:")
print(f"Obciążenie równomierne: {masa_calkowita * 9.81 / 1000 / (L*W):.2f} kN/m²")
print(f"Przy 8 punktach podparcia: {masa_calkowita * 9.81 / 1000 / 8:.2f} kN/punkt ({masa_calkowita/8:.0f} kg/punkt)")
```

### 3.5. Wizualizacja rozkładu masy

```python
# ============================================================
# WYKRES KOŁOWY - ROZKŁAD MASY
# ============================================================

import matplotlib.pyplot as plt

# Przygotowanie danych
labels = df["Kategoria"].tolist()
sizes = df["Masa [kg]"].tolist()
colors = plt.cm.Set3(range(len(labels)))

# Wykres kołowy
fig, ax = plt.subplots(figsize=(12, 8))
wedges, texts, autotexts = ax.pie(
    sizes, 
    labels=labels, 
    autopct='%1.1f%%',
    colors=colors,
    startangle=90,
    textprops={'fontsize': 9}
)

# Poprawa czytelności
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_weight('bold')

ax.set_title(f'Rozkład masy całkowitej budynku\n({masa_calkowita:.0f} kg = {masa_calkowita/1000:.2f} ton)', 
             fontsize=14, weight='bold')

plt.tight_layout()
plt.show()

print("Wykres przedstawia procentowy udział poszczególnych elementów w masie całkowitej budynku.")
```
